### 중복 데이터 제거

```python
import pandas as pd
import glob

csv_files = glob.glob('data/*.csv')

dataframes = [pd.read_csv(file) for file in csv_files]
merged_df = pd.concat(dataframes, ignore_index=True)

deduplicated_df = merged_df.drop_duplicates(subset='address', keep='first')
deduplicated_df.to_csv('studyroom_data.csv', index=False)
```

### 스터디룸 필터링

filtering by. 스터디카페, 스터디룸, 워크, 스페이스, 공유오피스, 토즈

```python
keywords = ['스터디카페', '스터디룸', '워크', '스페이스', '공유오피스', '토즈', '스터디', '모임공간']

deduplicated_df['studyroom'] = deduplicated_df['name'].apply(
    lambda x: 'yes' if any(keyword in x for keyword in keywords) else 'no'
)
```

```python
study_yes = deduplicated_df[deduplicated_df['studyroom']=='yes']
# study_no = deduplicated_df[deduplicated_df['studyroom']=='no']
study_yes.to_csv('studyroom_yes.csv', index=False)
# study_no.to_csv('studyroom_no.csv', index=False)
```

### 구 기반 분류

address 칼럼의 구 정보 기반으로 분류

```python
study_yes['area'] = study_yes['address'].str.extract(r'(\S+구)')
```

```python
area = study_yes['area'].dropna().unique()
area
```

```python
study_yes.to_csv('studyroom_with_area.csv', index=False)
```

### 역정보 기반 분류

subway 칼럼의 근처역 정보 기반으로 역정보 분류

```python
study_yes.rename(columns={'subway': 'subway_distance'}, inplace=True)
study_yes['subway'] = study_yes['subway_distance'].str.extract(r'(\S+역)')
```

```python
area = study_yes['subway'].dropna().unique()
area # 총 104개 역
```

```python
import re

def clean_subway_distance(text):
    # 역 이름과 출구 번호, 거리를 추출하는 정규표현식
    pattern = r'(?:[\d\s]+)?(?:[가-힣]+\s+)?([가-힣()]+역)\s+(\d+번\s*출구)에서\s+(\d+)\s*(?:m|미터)'
    match = re.search(pattern, text)
    
    if match:
        station, exit_num, distance = match.groups()
        # 불필요한 공백 제거 및 형식 통일
        result = f"{station.strip()} {exit_num.replace(' ', '')}에서 {distance}m"
        return result
    
    return text  # 매치되지 않으면 원본 텍스트 반환


study_yes['subway_distance'] = study_yes['subway_distance'].apply(clean_subway_distance)
```

```python
study_yes.to_csv('studyroom_with_subway.csv')
```

### 리뷰, 별점 데이터 클리닝

+별점 정보: 방문자 리뷰의 별점 선택 기능이 종료되어 현재 표시되는 평균 별점은 2021년 10월까지 참여한 결과입니다.

```python
def split_reviews(row):
    # 별점 추출
    star_match = re.search(r'별점([\d.]+)', row['visitor_review'])
    star = star_match.group(1) if star_match else None
    
    # 방문자 리뷰 추출
    visitor_match = re.search(r'방문자 리뷰 (\d+)', row['visitor_review'])
    visitor = visitor_match.group(1) if visitor_match else None
    
    # 블로그 리뷰 추출
    blog_match = re.search(r'블로그 리뷰 (\d+)', row['blog_review'])
    blog = blog_match.group(1) if blog_match else None
    
    return pd.Series({'visitor_review': f'방문자 리뷰 {visitor}' if visitor else None,
                      'blog_review': f'블로그 리뷰 {blog}' if blog else None,
                      'star': f'별점 {star}' if star else None})

study_yes[['visitor_review', 'blog_review', 'star']] = study_yes.apply(split_reviews, axis=1)
```

```python
study_yes.to_csv('studyroom_with_review.csv')
```
